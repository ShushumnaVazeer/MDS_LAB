## Python List vs NumPy Array Performance Comparison

This section compares the performance of Python's built-in lists and NumPy arrays for common operations, specifically focusing on creation and element-wise arithmetic. NumPy arrays are generally more efficient for numerical operations, especially with large datasets, due to their underlying C implementation and optimized memory usage.

```python
import time
import numpy as np

# Define the size of the data structures
size = 10**6  # One million elements

print(f"Comparing performance for data structures with {size} elements.\n")

# --- 1. Creation Time ---
print("### 1. Creation Time ###")

# Python List creation
start_time = time.time()
python_list = list(range(size))
end_time = time.time()
python_list_creation_time = end_time - start_time
print(f"Python List Creation: {python_list_creation_time:.6f} seconds")

# NumPy Array creation
start_time = time.time()
numpy_array = np.arange(size)
end_time = time.time()
numpy_array_creation_time = end_time - start_time
print(f"NumPy Array Creation: {numpy_array_creation_time:.6f} seconds")

print(f"NumPy array creation is {python_list_creation_time / numpy_array_creation_time:.2f}x faster.\n")

# --- 2. Element-wise Addition ---
print("### 2. Element-wise Addition ###")

# Python List addition
start_time = time.time()
python_list_sum = [x + 1 for x in python_list]
end_time = time.time()
python_list_add_time = end_time - start_time
print(f"Python List Addition: {python_list_add_time:.6f} seconds")

# NumPy Array addition
start_time = time.time()
numpy_array_sum = numpy_array + 1
end_time = time.time()
numpy_array_add_time = end_time - start_time
print(f"NumPy Array Addition: {numpy_array_add_time:.6f} seconds")

print(f"NumPy array addition is {python_list_add_time / numpy_array_add_time:.2f}x faster.\n")


# --- 3. Element-wise Multiplication ---
print("### 3. Element-wise Multiplication ###")

# Python List multiplication
start_time = time.time()
python_list_prod = [x * 2 for x in python_list]
end_time = time.time()
python_list_mul_time = end_time - start_time
print(f"Python List Multiplication: {python_list_mul_time:.6f} seconds")

# NumPy Array multiplication
start_time = time.time()
numpy_array_prod = numpy_array * 2
end_time = time.time()
numpy_array_mul_time = end_time - start_time
print(f"NumPy Array Multiplication: {numpy_array_mul_time:.6f} seconds")

print(f"NumPy array multiplication is {python_list_mul_time / numpy_array_mul_time:.2f}x faster.\n")
```

```python
import sys

# --- 4. Memory Usage ---
print("### 4. Memory Usage ###")

# Python List memory usage
# sys.getsizeof() reports the size of the list object itself, plus a pointer for each item.
# For a more accurate comparison of actual data storage, we also need to consider the size of individual elements.
# However, for a simple comparison, sys.getsizeof() on the list object and numpy's .nbytes is often used.
python_list_memory = sys.getsizeof(python_list)
print(f"Python List Memory: {python_list_memory / (1024*1024):.3f} MB") # Convert bytes to MB

# NumPy Array memory usage
numpy_array_memory = numpy_array.nbytes
print(f"NumPy Array Memory: {numpy_array_memory / (1024*1024):.3f} MB") # Convert bytes to MB

if numpy_array_memory > 0:
    print(f"NumPy array uses {python_list_memory / numpy_array_memory:.2f}x less memory (relative to Python list object size and internal pointers).\n")
else:
    print("Cannot compare memory usage as NumPy array memory is zero (likely an empty array).\n")
```

## Conclusion

As demonstrated, NumPy arrays consistently outperform Python lists for numerical operations on large datasets. This is primarily due to:

*   **Vectorization**: NumPy operations are implemented in C, allowing for highly optimized, vectorized computations without explicit Python loops.
*   **Memory Efficiency**: NumPy arrays store elements of the same data type contiguously in memory, leading to better cache performance and reduced memory overhead compared to Python lists, which store pointers to objects.

For tasks involving significant numerical computation, using NumPy is highly recommended to achieve better performance.

```python
import numpy as np
import builtins

# Define a dictionary
data_dict = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

print("Original Dictionary:", data_dict)

# Create a NumPy array from dictionary values
array_from_values = np.array(builtins.list(data_dict.values()))
print("\nNumPy Array from Dictionary Values:", array_from_values)
print("Type:", type(array_from_values))

# Create a NumPy array from dictionary keys
array_from_keys = np.array(builtins.list(data_dict.keys()))
print("\nNumPy Array from Dictionary Keys:", array_from_keys)
print("Type:", type(array_from_keys))

# Create a NumPy array from dictionary items (list of tuples)
array_from_items = np.array(builtins.list(data_dict.items()))
print("\nNumPy Array from Dictionary Items (as list of tuples):\n", array_from_items)
print("Type:", type(array_from_items))
```
