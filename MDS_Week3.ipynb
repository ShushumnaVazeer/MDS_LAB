```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
# Read image
image = io.imread("/content/cat image.jpg")
# Crop both eyes, excluding the nose
eyes = image[1450:2000, 700:2400] # Adjusted coordinates to exclude nose
# Display original image
plt.figure(figsize=(6,10))
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()
# Display cropped eyes
plt.figure(figsize=(8,4))
plt.imshow(eyes)
plt.title("Both Eyes") # Updated title
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
# Load image
image = io.imread("/content/cat image.jpg")
# Rotate images
rotate_90 = np.rot90(image)
rotate_180 = np.rot90(image, 2)
rotate_270 = np.rot90(image, 3)
# Flip images
horizontal_flip = image[:, ::-1]
vertical_flip = image[::-1, :]
# Display using subplots
fig, ax = plt.subplots(2, 3, figsize=(12, 10))
ax[0, 0].imshow(image)
ax[0, 0].set_title("Original")
ax[0, 0].axis("off")
ax[0, 1].imshow(rotate_90)
ax[0, 1].set_title("90° Rotation")
ax[0, 1].axis("off")
ax[0, 2].imshow(rotate_180)
ax[0, 2].set_title("180° Rotation")
ax[0, 2].axis("off")
ax[1, 0].imshow(rotate_270)
ax[1, 0].set_title("270° Rotation")
ax[1, 0].axis("off")
ax[1, 1].imshow(horizontal_flip)
ax[1, 1].set_title("Horizontal Flip")
ax[1, 1].axis("off")
ax[1, 2].imshow(vertical_flip)
ax[1, 2].set_title("Vertical Flip")
ax[1, 2].axis("off")
plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
# Read the image using cv2.imread (it reads in BGR format by default)
bgr_image = cv2.imread("/content/cat image.jpg")
# Convert BGR to RGB
rgb_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)
# Display both images
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(bgr_image) # Matplotlib expects RGB, so this will appear discolored as BGR
ax[0].set_title("Original Image (BGR format by cv2.imread)")
ax[0].axis("off")
ax[1].imshow(rgb_image)
ax[1].set_title("Converted to RGB")
ax[1].axis("off")
plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Import numeric columns from Iris CSV
data = np.loadtxt("/content/iris.csv",
                  delimiter=",",
                  skiprows=1,
                  usecols=(0,1,2,3))



# ---------- Statistical Operations ----------

print("\nMean:")
print(np.mean(data, axis=0))

print("\nMedian:")
print(np.median(data, axis=0))

print("\nMaximum:")
print(np.max(data, axis=0))

print("\nMinimum:")
print(np.min(data, axis=0))

print("\nStandard Deviation:")
print(np.std(data, axis=0))

# ---------- Comparison Operations ----------

print("\nRows where Sepal Length > 6.0")
print(data[data[:,0] > 6.0])

print("\nRows where Sepal Width > 3.5")
print(data[data[:,1] > 3.5])

print("\nRows where Petal Length >= 5.0")
print(data[data[:,2] >= 5.0])

print("\nRows where Petal Width < 1.0")
print(data[data[:,3] < 1.0])
```
